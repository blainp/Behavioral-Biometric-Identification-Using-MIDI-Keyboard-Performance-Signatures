```python
import pandas as pd

# List of your files in order
files = [
    "features0.csv",
    "features1.csv",
    "features2.csv",
    "features3.csv",
    "features4.csv",
    "features5.csv"
]

dataframes = []
user_id = 0

for i, file in enumerate(files):
    df = pd.read_csv(file)
    
    # Assign user ID
    df.insert(0, "userID", user_id)
    
    dataframes.append(df)
    
    # Increment user_id EXCEPT for features4 and features5
    if i < 4:   # features0 to features3 → increment normally
        user_id += 1
    elif i == 4:
        pass    # features4 → do NOT increment yet (same as next)
    elif i == 5:
        user_id += 1  # increment after features5 if needed later

# Stack all datasets vertically
stacked_df = pd.concat(dataframes, ignore_index=True)

# Save result
stacked_df.to_csv("stacked_dataset.csv", index=False)

# Preview
stacked_df.head()
```

```python
#Remove non-entry values and replace them with 0s
import pandas as pd

# Load dataset
df = pd.read_csv('stacked_dataset.csv')

# Convert non-numeric to NaN, then replace with 0
df = df.apply(pd.to_numeric, errors='coerce').fillna(0)

# Save back to same file
df.to_csv('stacked_dataset.csv', index=False)
```
